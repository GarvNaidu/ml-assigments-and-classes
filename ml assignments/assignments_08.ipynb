```python
import pandas as pd
df = {
    'House': ['A', 'B', 'C', 'D'],
    'Area': [1200, 1800, 2400, 1500],
    'Bedrooms': [2, 3, 4, 3],
    'Age': [10, 5, 8, 12],
    'Location Score': [3, 4, 5, 2]
}
df01 = pd.DataFrame(df)

df01['Predicted Price'] = (12 
+ 0.045 * df01['Area']  
+ 6 * df01['Bedrooms'] 
- 1.2 * df01['Age'] 
+ 8 * df01['Location Score'])

df01[['House', 'Predicted Price']]
```

*Interpretation of Feature Coefficients*

**1- Area: For every additional square unit of area, the predicted price increases by $0.045 units**

**2- Bedrooms: For every additional bedroom, the predicted price increases by $6.0 units**

**3- Age: For every additional year of age (older house), the predicted price decreases by $1.2 units**

**4- Location Score: For every 1-unit increase in the location score, the predicted price increases by $8.0 units**
