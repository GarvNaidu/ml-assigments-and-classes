```python
import pandas as pd

df01 = {
    "A": {"Area": 1500, "Bedrooms": 3, "Age": 5},
    "B": {"Area": 2200, "Bedrooms": 4, "Age": 10},
    "C": {"Area": 1000, "Bedrooms": 2, "Age": 15}
}

df = pd.DataFrame.from_dict(df01, orient='index')

df['Predicted_Price'] = 15 + (0.08 * df['Area']) + (5 * df['Bedrooms']) - (0.6 * df['Age'])

df[['Predicted_Price']]
```

*An automobile company builds a multiple linear regression model to predict used-car prices using vehicle age, mileage, and engine capacity. After training, the company observes that the model has a high \(R^2\) score but still makes large errors for expensive cars.*

*Tasks:*
*Explain how coefficients and the intercept determine the model's predictions, and identify a limitation of interpreting coefficients when input features are correlated.*

**Intercept :** The intercept gives the starting price of the house when all the input values are zero.

**Coefficients:** Each coefficient tells us how much the house price changes when one feature increases by 1 unit, while keeping the other features unchanged.

**Final Prediction:** To predict the price, we multiply each feature by its respective coefficient and add all the results with the intercept.

**Correlation Limitation:** Sometimes, two features are closely related, such as the age of a vehicle and its mileage. In such cases, it is difficult to change one feature without affecting the other.

**Unreliable Values:** When features are highly correlated, the model may struggle to identify their individual effects. This can make the coefficients unstable and less reliable.

PART 2 - Regression intuition and line

```python
from IPython.core.display_functions import display
import numpy as np
df02 = {
    'Student': ['A', 'B', 'C', 'D', 'E', 'F'],
    'Study Hours (x)': [1, 2, 4, 5, 7, 8],
    'Actual Marks (y)': [3, 4, 7, 8, 12, 10]
}
df03 = pd.DataFrame(df02)

x = df03['Study Hours (x)'].values
y = df03['Actual Marks (y)'].values

mean_x = np.mean(x)
mean_y = np.mean(y)

numerator = np.sum((x - mean_x) * (y - mean_y))
denominator = np.sum((x - mean_x) ** 2)
slope = numerator / denominator

intercept = mean_y - slope * mean_x

print("mean of x:",mean_x)
print("mean of y:",mean_y)
print("ols slope:",slope)
print("intercept:", intercept)

print(f"Regression Equation: y = {slope}x + {intercept}\n")
print("----------------------------------------------------------------------")
print("Predicted marks and residuals for all six students")
predicted_marks = slope * x + intercept
residuals = y - predicted_marks
df03['predicted_marks'] = predicted_marks
df03['residuals'] = residuals
print("predicted marks:", predicted_marks)
print("residuals:", residuals)
print("----------------------------------------------------------------------")
sse = np.sum(df03['residuals'] ** 2)
mse = sse / len(df03)

print("Final DataFrame Summary")
df03
```

```python
import pandas as pd
import numpy as np

x = np.array([2, 4, 6, 8, 10, 12], dtype=float)
y = np.array([15, 20, 28, 35, 43, 48], dtype=float)

mean_x01 = np.mean(x)
mean_y01 = np.mean(y)

num01 = np.sum((x - mean_x01) * (y - mean_y01))
den01 = np.sum((x - mean_x01) ** 2)

slope01 = num01 / den01
intercept01 = mean_y01 - (slope01 * mean_x01)

print(f"Slope: {slope01:.2f}")
print(f"Intercept: {intercept01:.2f}")

print("-" * 30)

x_new = np.array([5, 9, 16], dtype=float)
y_new_pred = slope01 * x_new + intercept01

for xi, yi in zip(x_new, y_new_pred):
    estimation = "interpolation" if 2 <= xi <= 12 else "extrapolation"
    print(f"x = {xi}: y pred = {yi:.2f} = {estimation}")

print("-" * 30)

y_pred = slope01 * x + intercept01
residuals02 = y - y_pred
squared_residuals = residuals02 ** 2

sse = np.sum(squared_residuals)
mse = np.mean(squared_residuals)
rmse = np.sqrt(mse)

print(f"SSE: {sse:.2f}")
print(f"MSE: {mse:.2f}")
print(f"RMSE: {rmse:.2f}")
df04 = pd.DataFrame({
    'x': x,
    'Actual y': y,
    'Predicted y': y_pred,
    'Residual': residuals02,
    'Squared Residual': squared_residuals
})

df04
```

```python

intercept = 25
slope = 1.8
operating_hours = [40, 85, 120, 200]

min = 30
max = 100

data = []
for hours in operating_hours:
    predicted_fuel = intercept + (slope * hours)
    
    if min <= hours <= max:
        classification = "interpolation"
    else:
        classification = "extrapolation"
        
    data.append({
        "Operating Hours": hours,
        "Predicted Fuel Consumption": predicted_fuel,
        "Classification": classification
    })

df05 = pd.DataFrame(data)

df05
```
