```python
import pandas as pd
df = {
    'model': ['A','B','C','D'],
    'max_depth': [1,4,8,15],
    'training accuracy': [68, 87, 98, 100],
    'validation accuracy': [66,84,76,69]
}
df01 = pd.DataFrame(df)

def classify_model(x):
    train = x['training accuracy']
    val = x['validation accuracy']
    gap = train - val

    if train < 75 and val < 75:
        return 'Underfitting'
    elif gap <= 5:
        return 'Good Fit'
    else:
        return 'Overfitting'

df01['fit status'] = df01.apply(classify_model, axis = 1)

df01['accuracy gap'] = df01['training accuracy'] - df01['validation accuracy']

df01
```
